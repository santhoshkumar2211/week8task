```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = sns.load_dataset('tips')

print(df)
```

```python
print(df.head())
print(df.tail())
print("Shape:", df.shape)
print("Columns:", df.columns)
print(df.dtypes)
print("Missing Values:")
print(df.isnull().sum())
df.info()
```

```python
for col in df.select_dtypes(include=np.number).columns:
    df[col] = df[col].fillna(df[col].median())
for col in df.select_dtypes(include='category').columns:
    df[col] = df[col].fillna(df[col].mode()[0])
print("Missing values after handling:")
print(df.isnull().sum())
```

```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['sex_encoded'] = le.fit_transform(df['sex'])
df['smoker_encoded'] = le.fit_transform(df['smoker'])
df['day_encoded'] = le.fit_transform(df['day'])
df['time_encoded'] = le.fit_transform(df['time'])

print(df.head())
```

```python
categorical_columns = ['sex', 'smoker', 'day', 'time']

for col in categorical_columns:
    print("\nFrequency Distribution:", col)
    print(df[col].value_counts())
```

```python
for col in categorical_columns:
    print("\nPercentage Distribution:", col)
    print(df[col].value_counts(normalize=True) * 100)
```

```python
for col in categorical_columns:
    print("Most Common Category in", col, ":", df[col].mode()[0])
```

```python
plt.figure(figsize=(7, 5))

sns.countplot(data=df, x='day')

plt.title("Count Plot - Day Distribution")
plt.xlabel("Day")
plt.ylabel("Count")

plt.show()
```

```python
sex_counts = df['sex'].value_counts()

plt.figure(figsize=(6, 6))

plt.pie(
    sex_counts,
    labels=sex_counts.index,
    autopct='%1.1f%%',
    startangle=90
)

plt.title("Gender Distribution")

plt.show()
```

```python
smoker_counts = df['smoker'].value_counts()

plt.figure(figsize=(7, 5))

plt.bar(smoker_counts.index.astype(str), smoker_counts.values)

plt.title("Smoker Distribution")
plt.xlabel("Smoker")
plt.ylabel("Count")

plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.countplot(data=df, x='day', hue='sex')

plt.title("Day Distribution by Gender")
plt.xlabel("Day")
plt.ylabel("Count")

plt.show()
```
